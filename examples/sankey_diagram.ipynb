```python
import plotly.graph_objects as go
import pandas as pd

# Define the nodes as specified
labels = [
    "Products Revenue",       # 0
    "Services Revenue",       # 1
    "Net sales",              # 2
    "Cost of sales (Products)", # 3
    "Cost of sales (Services)", # 4
    "Total Cost of Sales",    # 5
    "Gross margin",           # 6
    "R&D",                    # 7
    "SG&A",                   # 8
    "Operating income",       # 9
    "Other income",           # 10
    "Income before taxes",    # 11
    "Income taxes",           # 12
    "Net income"              # 13
]

# Define node x-positions (column organization)
# Scale back the positions a bit to ensure the rightmost nodes are fully visible
node_x = [
    0.001,  # "Products Revenue" - column 1
    0.001,  # "Services Revenue" - column 1
    0.2,    # "Net sales" - column 2
    0.6,    # "Cost of sales (Products)" - column 4 (same as Operating income)
    0.6,    # "Cost of sales (Services)" - column 4 (same as Operating income)
    0.4,    # "Total Cost of Sales" - column 3  
    0.4,    # "Gross margin" - column 3
    0.6,    # "R&D" - column 4
    0.6,    # "SG&A" - column 4
    0.6,    # "Operating income" - column 4
    0.6,    # "Other income" - column 4
    0.8,    # "Income before taxes" - column 5 (moved left)
    1.0,    # "Income taxes" - column 6 (moved left)
    1.0     # "Net income" - column 6 (moved left)
]

# Define node y-positions (vertical placement within columns)
node_y = [
    0.2,    # "Products Revenue" 
    0.8,    # "Services Revenue"
    0.5,    # "Net sales"
    0.2,    # "Cost of sales (Products)" 
    0.35,   # "Cost of sales (Services)"
    0.5,    # "Total Cost of Sales"
    0.8,    # "Gross margin"
    0.7,    # "R&D"
    0.85,   # "SG&A"
    0.5,    # "Operating income"
    0.15,   # "Other income"
    0.5,    # "Income before taxes"
    0.2,    # "Income taxes"
    0.8     # "Net income"
]

# Define the connections (source → target) as specified
source = [
    0, 1,              # Revenue to Net sales
    2, 2,              # Net sales to total cost of sales and gross margin
    5, 5,              # Total cost of sales to individual components
    6, 6, 6,           # Gross margin to R&D, SG&A and operating income
    9, 10,             # Operating income and other income to income before taxes
    11, 11             # Income before taxes to taxes and net income
]

target = [
    2, 2,              # Revenue to Net sales
    5, 6,              # Net sales to total cost of sales and gross margin
    3, 4,              # Total cost of sales to individual components
    7, 8, 9,           # Gross margin to R&D, SG&A and operating income
    11, 11,            # Operating income and other income to income before taxes
    12, 13             # Income before taxes to taxes and net income
]

# Define the values (flow amounts) based on the original data
value = [
    # Revenue to Net sales
    220747, 53768,
    
    # Net sales to total cost of sales and gross margin
    169559, 104956,
    
    # Total cost of sales to individual components
    151286, 18273,
    
    # Gross margin to expenses and operating income
    18752, 19916, 66288,
    
    # Operating income and other income to income before taxes
    66288, 803,
    
    # Income before taxes to taxes and net income
    9680, 57411
]

# Create the Sankey diagram with explicit node positions
fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=labels,
        x=node_x,  # Set explicit x positions
        y=node_y,  # Set explicit y positions
        color="lightblue"
    ),
    link=dict(
        source=source,
        target=target,
        value=value,
        hovertemplate='$%{value} million<extra></extra>',
        color="rgba(100, 100, 100, 0.4)"
    ),
    # Use "fixed" arrangement to respect our manual positioning
   # arrangement="fixed"
)])

# Update the layout with better styling and increased margins
fig.update_layout(
    title_text="Financial Flow (in millions USD)",
    font_size=12,
    height=700,
    width=1000,  # Explicitly set width to ensure enough space
    # Use a nice color palette for nodes
    colorway=["#5e4fa2", "#3288bd", "#66c2a5", "#abdda4", "#e6f598", 
              "#fee08b", "#fdae61", "#f46d43", "#d53e4f", "#9e0142"],
    # Add generous margins, especially on the right
    margin=dict(t=60, l=25, r=150, b=25)  # Increased right margin
)

# Show the diagram
fig.show()
```
